```python
from pyspark.sql.functions import count
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

silver_path = "/Volumes/brewery/silver_db/breweries_parquet"

df_silver = spark.read.parquet(silver_path)

df_gold = (
    df_silver
    .groupBy("country", "state_province", "brewery_type")
    .agg(count("*").alias("brewery_count"))
)

gold_path = "/Volumes/brewery/gold_db/breweries_aggregated"

df_gold.write.mode("overwrite").partitionBy("country", "state_province").parquet(gold_path)
```
